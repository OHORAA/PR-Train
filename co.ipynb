```python
fig, ax = plt.subplots(2,2)

x = np.arange(5)

ax[0,0].plot(x,x)
ax[0,0].set_title("y = x")

ax[0,1].bar(x,2*x)
ax[0,1].set_title("y = 2x")

ax[1,0].scatter(x,x*x)
ax[1,0].set_title("y = x^2")

ax[1,1].hist(x,x**3)
ax[1,1].set_title("y = x^3")



fig.suptitle('subplot graph', fontsize=16)
plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(2,2)

for i, axi in enumerate(ax.flat):
    axi.plot([1,2,3],[n*i+1 for n in [1,2,3]])
    axi.set_title(f'Plot {i+1}')

plt.tight_layout()
plt.show()
```
